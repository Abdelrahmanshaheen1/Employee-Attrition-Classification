```python
# C Apply Python techniques to create new features from existing ones (feature engineering) and explain the significance of the new features.


import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)  # Show all columns
pd.set_option('display.width', 1000)  # Expand display width

# Feature Engineering

# Measures productivity the higher it is the good it is
dataset2["Work_Hours_Efficiency"] = dataset2["Project_Count"] / dataset2["Average_Hours_Worked_Per_Week"]

# we will make age as group like from 18to25 kda yaany kda the manger for example will now which age group are the most common
bins = [18, 25, 35, 45, 55, 65]
labels = ["18-25", "26-35", "36-45", "46-55", "56-65"]
dataset2["Age_Group"] = pd.cut(dataset2["Age"], bins=bins, labels=labels)

#  we wiil change yesto1 and no to 0 Helps in workload analysis
dataset2["Overtime_Binary"] = dataset2["Overtime"].map({"Yes": 1, "No": 0})

# usfull because you will know who should get promation for example and aslo help you to know the growth potential of an employee
dataset2["Experience_Score"] = (dataset2["Years_at_Company"] + dataset2["Years_in_Current_Role"]) / 2

# basicaly thsi calculates how often an employee gets a promation ,Helps in succession planning and evaluating internal growth, so for a employee he will hope that it will be high
dataset2["Promotion_Frequency"] = dataset2["Years_at_Company"] / (dataset2["Years_Since_Last_Promotion"] + 1)

#helps to see how happy or satesfied in his job
dataset2["Job_Satisfaction_Score"] = (
    dataset2["Job_Satisfaction"] + dataset2["Work_Environment_Satisfaction"] + dataset2["Relationship_with_Manager"]
) / 3

#betkhalyk tshof how stable is the employee career and how loyal he or she is for the company closer to 1 means that he is loyal
dataset2["Career_Stability"] = (dataset2["Years_at_Company"] - dataset2["Number_of_Companies_Worked"]) / dataset2["Years_at_Company"]

#see how often an employee gets promated
dataset2["Promotion_Rate"] = dataset2["Years_Since_Last_Promotion"] / dataset2["Years_at_Company"]

#helps you to know who shoild get a promation
dataset2["Promotion_Opportunity"] = dataset2["Years_Since_Last_Promotion"] / (dataset2["Years_at_Company"] + 1)


#Measures workload stress ,usful fro workforce balancing and improving efficiency.
dataset2["Workload_Score"] = (dataset2["Overtime_Binary"] * dataset2["Project_Count"]) / dataset2["Average_Hours_Worked_Per_Week"]


#helps you to see who would likely to leave the compony ,so you could possibly solve his problems
dataset2["Attrition_Risk"] = (1 - dataset2["Job_Satisfaction_Score"]) * dataset2["Workload_Score"]


# print tany to see the new added columns
print(dataset2.head(5))


# make sure the Attrition_Risk is in postive value
dataset2["Attrition_Risk"] = dataset2["Attrition_Risk"].abs()


plt.figure(figsize=(12, 5))#size of the figure

# Attrition Risk by Job Role
plt.subplot(1, 2, 1)
#Plots the average attrition risk for each job role.
sns.barplot(data=dataset2, x="Job_Role", y="Attrition_Risk", hue="Job_Role", palette="coolwarm", dodge=False)
plt.xticks(rotation=90)
plt.xlabel("Job Role")#xaxis
plt.ylabel("Avg Attrition Risk")#yaxis
plt.title("Attrition Risk by Job Role")#title
plt.legend([], [], frameon=False)

# Attrition Risk by Department
plt.subplot(1, 2, 2)
sns.barplot(data=dataset2, x="Department", y="Attrition_Risk", hue="Department", palette="viridis", dodge=False)
plt.xticks(rotation=45)
plt.xlabel("Department")
plt.ylabel("Avg Attrition Risk")
plt.title("Attrition Risk by Department")
plt.legend([], [], frameon=False)
plt.tight_layout()
plt.show()











import seaborn as sns
import matplotlib.pyplot as plt
#Histogram: Shows overall distribution.
#Boxplot: Displays spread, median, and quartiles with key statistics#Also the mean i forgot to write it
#Barplot: Compares Promotion Frequency by Department & Job Role, i can change it to whatever i want here it was for the Promotion Frequency by Department & Job Role


Q1 = dataset2["Promotion_Opportunity"].quantile(0.25)#calculates the q1 for Promotion_Opportunity
Q3 = dataset2["Promotion_Opportunity"].quantile(0.75)#calculates the q3 for Promotion_Opportunity
median = dataset2["Promotion_Opportunity"].median()#calculates the median for Promotion_Opportunity
mean = dataset2["Promotion_Opportunity"].mean()#calculates the mean for Promotion_Opportunity
min_val = dataset2["Promotion_Opportunity"].min()#calculates the min for Promotion_Opportunity
max_val = dataset2["Promotion_Opportunity"].max()#calculates the max for Promotion_Opportunity

fig = plt.figure(figsize=(18, 5))  #the size of each one
axes = fig.subplots(1, 3) #divie it to 3 because w will have 3 graphs


# Histogram
#histplot(): Plots the distribution using 30 bins.
#kde=True: Adds a Kernel Density Estimation (KDE) curve for smooth visualization.
#ax=axes[0]: Assigns this plot to the first subplot.
#set_title(): Adds a title.
sns.histplot(dataset2["Promotion_Opportunity"], bins=30, kde=True, ax=axes[0])
axes[0].set_title("Distribution of Promotion Opportunity")


# Boxplot
sns.boxplot(x=dataset2["Promotion_Opportunity"], ax=axes[1], color="lightblue")
axes[1].set_title("Boxplot of Promotion Opportunity")

# Add statistics to the boxplot
axes[1].axvline(Q1, color="blue", linestyle="--", label=f"Q1: {Q1:.2f}")
axes[1].axvline(Q3, color="blue", linestyle="--", label=f"Q3: {Q3:.2f}")
axes[1].axvline(median, color="red", linestyle="-", label=f"Median: {median:.2f}")
axes[1].axvline(mean, color="green", linestyle="-.", label=f"Mean: {mean:.2f}")
axes[1].axvline(min_val, color="purple", linestyle=":", label=f"Min: {min_val:.2f}")
axes[1].axvline(max_val, color="purple", linestyle=":", label=f"Max: {max_val:.2f}")
axes[1].legend()


#barplot
sns.barplot(data=dataset2, x="Department", y="Promotion_Opportunity", hue="Job_Role", palette="viridis", ax=axes[2])
axes[2].set_title("Promotion Opportunity by Department & Job Role")

#Prevents overlapping labels.
axes[2].tick_params(axis="x", rotation=45)

plt.tight_layout()#byt adjust L Spaces
plt.show()

```

```python
#deeplearning

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix




```

```python
X = dataset2.drop('Attrition', axis=1)
y = dataset2['Attrition']
```

```python
categorical_cols = X.select_dtypes(include=['object']).columns
label_encoder = LabelEncoder()
for col in categorical_cols:
    X[col] = label_encoder.fit_transform(X[col])
```

```python
numerical_cols = X.select_dtypes(include=['number']).columns
scaler = StandardScaler()
X[numerical_cols] = scaler.fit_transform(X[numerical_cols])
```

```python
y_encoder = LabelEncoder()
y_encoded = y_encoder.fit_transform(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)
```

```python
# Define the deep learning model
model = keras.Sequential([
    layers.Input(shape=(X_train.shape[1],)),
    layers.Dense(64, activation='relu'),
    layers.Dense(32, activation='relu'),
    layers.Dense(1, activation='sigmoid')  # Output layer for binary classification
])
# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])



# Train the model
model.fit(X_train, y_train, epochs=10, batch_size=32)
```

```python
y_pred_prob = model.predict(X_test)
y_pred_dl = (y_pred_prob > 0.5).astype(int)
```

```python
accuracy_dl = accuracy_score(y_test, y_pred_dl)
precision_dl = precision_score(y_test, y_pred_dl, pos_label=y_encoder.transform(['Yes'])[0])  # Use encoded label for 'Yes'
recall_dl = recall_score(y_test, y_pred_dl, pos_label=y_encoder.transform(['Yes'])[0])  # Use encoded label for 'Yes'
cm_dl = confusion_matrix(y_test, y_pred_dl)

# Print the results
print("Deep Learning Model:")
print("Accuracy:", accuracy_dl)
print("Confusion Matrix:\n", cm_dl)
print("Precision:", precision_dl)
print("Recall:", recall_dl)
```
